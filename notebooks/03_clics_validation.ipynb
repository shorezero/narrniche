## Shared Lexicon between Fields and CLICS

```python
def gloss(node, graph):
    return graph.nodes[node]['Gloss'].lower().split(' (')[0]
```

## CLICS Clusters

### Infomap

```python
# clics infomap fields
clics_im = dict()
for n in clics.nodes:
    c = clics.nodes[n]['infomap']
    g = gloss(n, clics)
    clics_im[c] = clics_im[c] + [g] if c in clics_im else [g]
print(len(clics_im))
clics_im
```

### k-Cliques

```python
# clics k-cliques + arcs fields
clics_kc = list(map(list, nx.community.k_clique_communities(clics, 3)))
present = {n for f in clics_kc for n in f}
for (a,b) in clics.edges:
    if a not in present and b not in present:
        if [b,a] not in clics_kc:
            clics_kc.append([a,b])
clics_kc = {i:[gloss(n, clics) for n in field] for i,field in enumerate(clics_kc)}
print(len(clics_kc))
clics_kc
```

# Edge Validation

```python
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score
```

```python
# load colexifications and clics
colex = nx.Graph(dill.load(open('../data/processed/colex.pkl', 'rb')))
clics = nx.parse_gml(open("../data/graphs/infomap-3-families.gml").read(), destringizer=int)
```

```python
# use subgraphs of shared lexicon
acolex = colex.subgraph([n for n in colex.nodes if lexicon[n] in alex])
aclics = clics.subgraph([n for n in clics.nodes if gloss(n, clics) in alex])
```

```python
e2s = lambda x: x[0]+'_'+x[1]
acolex_edges = sorted([e2s(sorted((lexicon[i], lexicon[j]))) for i,j in acolex.edges])
aclics_edges = sorted([e2s(sorted((gloss(i, aclics), gloss(j, aclics)))) for i,j in aclics.edges])
```

```python
a = set(acolex_edges)
b = set(aclics_edges)
```
